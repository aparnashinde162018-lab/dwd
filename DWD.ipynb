```python
import pandas as pd
```

```python
df=pd.DataFrame({"Hours_studied":[1,2,3,4,5,6,7,8,9],"Test_score":[15,55,18,70,65,75,55,80,85]})
df
```

```python
df.describe()
```

```python
mean_hours=df["Hours_studied"].mean()
mean_hours
```

```python
mean_score=df["Test_score"].mean()
mean_score
```

```python
median_hours=df["Hours_studied"].median()
median_hours
```

```python
median_score=df["Test_score"].median()
median_score
```

```python
mode_hours=df["Hours_studied"].mode().iloc[2]
mode_hours
```

```python
mode_score=df["Test_score"].mode().iloc[0]
mode_score
```
